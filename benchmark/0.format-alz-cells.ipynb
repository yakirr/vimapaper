```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
import scanpy as sc 
import anndata as ad 
```

```python
# remove duplicate cells
datadir = '../../ALZ/alz-data'
d = sc.read_h5ad(f'{datadir}/SEAAD_MTG_MERFISH.2024-12-11.h5ad')
d = d[~d.obs.duplicated(subset=['Specimen Barcode', 'Cell ID'], keep='first')]
d.obs.rename(columns={'Donor ID':'donor', 'Specimen Barcode':'sid'}, inplace=True)
d.obs['cid'] = d.obs.donor.astype(str) + '_' + d.obs.sid.astype(str) + '_' + d.obs['Cell ID'].astype(str)
d.obs.set_index('cid', inplace=True, drop=True)
d.write(f'{datadir}/SEAAD_MTG_MERFISH.2024-12-11.filtered.h5ad')
```
