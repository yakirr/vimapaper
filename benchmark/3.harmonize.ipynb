```python
import vima
import scanpy as sc
import glob, os
```

```python
def harmonize(dsetname, sid_to_covs=None):
    embeddings = glob.glob(f'_embeddings/{dsetname}*_noharm.h5ad')
    for embedding in embeddings:
        prefix = '_'.join(embedding.split('_')[:-1])
        print(prefix)

        if os.path.exists(f'{prefix}_harm.h5ad'):
            print('harmonized embedding already exists, skipping...')
        else:
            print('harmonizing...')
            adata = sc.read_h5ad(embedding)
            df = adata.obs.copy()
            if 'X_pca' not in adata.obsm:
                sc.pp.pca(adata)
            pcs = [f'PC{i+1}' for i in range(adata.obsm['X_pca'].shape[1])]
            print(len(pcs), 'PCs')
            for i, pc in enumerate(adata.obsm['X_pca'].T):
                df[f'PC{i+1}'] = pc
            harm_df = vima.pp.harmonize(df,
                                        '/Users/yakir/miniconda3/envs/Renv/bin/Rscript',
                                        sid_to_covs=sid_to_covs,
                                        plot=False)
            adata.obsm['X_pca'] = harm_df[pcs].values
            print('computing NN/UMAP')
            sc.pp.neighbors(adata, use_rep='X_pca')
            sc.tl.umap(adata)
            for clustering in [c for c in adata.obs.columns if 'leiden' in c]:
                resolution = float(clustering.split('_')[-1])
                print('\treclustering at resolution', resolution)
                sc.tl.leiden(adata, resolution=resolution, key_added=clustering)
            print('writing output')
            adata.write_h5ad(f'{prefix}_harm.h5ad')
            print('done')
```

```python
harmonize('ALZ')
```
