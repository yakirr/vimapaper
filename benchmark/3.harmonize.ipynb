```python
import vima
import scanpy as sc
import glob, os
```

```python
# where the embedding is stored for each method
reps = {
    'canvas':'X',
    'stagate':'STAGATE',
    'utag':'X',
    'cellcharter':'X_cellcharter',
    'tissuemosaic':'X',
    'patchavgmm':'X',
    'patchcelltypeabundance':'X',
}
```

```python
def harmonize(dsetname, sid_to_covs=None):
    embeddings = glob.glob(f'_embeddings/{dsetname}*_noharm.h5ad')
    for embedding in embeddings:
        prefix = '_'.join(embedding.split('_')[:-1])
        method = embedding.split('_')[-2]
        print(prefix, method)

        if os.path.exists(f'{prefix}_harm.h5ad'):
            print('harmonized embedding already exists, skipping...')
        else:
            print('harmonizing...')
            adata = sc.read_h5ad(embedding)
            df = adata.obs.copy()
            myrep = sc.AnnData(X=adata.X if reps[method] == 'X' else adata.obsm[reps[method]])
            sc.pp.pca(myrep, zero_center=True)
            pcs = [f'PC{i+1}' for i in range(myrep.obsm['X_pca'].shape[1])]
            print(len(pcs), 'PCs')
            for i, pc in enumerate(myrep.obsm['X_pca'].T):
                df[f'PC{i+1}'] = pc
            harm_df = vima.pp.harmonize(df,
                                        '/Users/yakir/miniconda3/envs/Renv/bin/Rscript',
                                        sid_to_covs=sid_to_covs,
                                        plot=False)
            adata.obsm['X_pca'] = harm_df[pcs].values
            print('computing NN/UMAP')
            sc.pp.neighbors(adata, use_rep='X_pca')
            sc.tl.umap(adata)
            for clustering in [c for c in adata.obs.columns if 'leiden' in c]:
                if clustering == 'leiden1':
                    resolution = 1
                else:
                    resolution = float(clustering.split('_')[-1])
                print('\treclustering at resolution', resolution)
                sc.tl.leiden(adata, resolution=resolution, key_added=clustering, flavor="igraph", n_iterations=2)
            print('writing output')
            adata.write_h5ad(f'{prefix}_harm.h5ad')
            print('done')
```

```python
harmonize('ALZ')
```

```python
harmonize('UC')
```

```python
harmonize('RA')
```
