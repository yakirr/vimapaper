```python
import os
import numpy as np
import pandas as pd
import scanpy as sc
```

```python
def run_patchavgs(dsetname):
    for embedding in ['avgmm', 'celltypeabundance']:
        fname = f'_data/{dsetname}/patches_{embedding}.h5ad'
        if os.path.exists(fname):
            d = sc.read_h5ad(fname)
            print(d.shape)
            d.write(f'_embeddings/{dsetname}_patch{embedding}_noharm.h5ad')
```

```python
run_patchavgs('RA')
run_patchavgs('UC')
run_patchavgs('ALZ')
```
